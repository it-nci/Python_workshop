### 1.1 📦 ตัวแปร และชนิดของข้อมูล

* **VARIABLES คืออะไร?**
ตัวแปร (Variables) เปรียบเสมือนกล่องติดป้ายชื่อที่ใช้สำหรับจัดเก็บข้อมูล โดยเราจะตั้งชื่อให้กล่องนั้นและใส่ข้อมูลลงไปข้างใน
* **ทำไมเราถึงต้องใช้ตัวแปร?**
แทนที่เราจะพิมพ์คำว่า "45" ซ้ำๆ ไปทุกที่ เราจะเขียนว่า "age" (อายุ) แทน — ซึ่งช่วยให้โค้ดอ่านเข้าใจง่ายขึ้น และแก้ไขข้อมูลในภายหลังได้สะดวกขึ้น
* **ชนิดข้อมูลพื้นฐานใน PYTHON (COMMON DATA TYPES):**

| ชนิดข้อมูล (Type) | ตัวอย่าง (Example) | การใช้งาน (Used For) |
| --- | --- | --- |
| **string (str)** | `"John Doe"` | ข้อความ, ชื่อ, รหัสต่างๆ |
| **integer** | `45` | จำนวนเต็ม, การนับจำนวน |
| **float** | `36.5` | ทศนิยม, ค่าจากการวัด |
| **boolean** | `True` / `False` | ค่าจริง/เท็จ, ใช่/ไม่ใช่, เปิด/ปิด |

* **ตัวอย่าง:**
เวลาที่คุณกรอกแบบฟอร์มข้อมูลคนไข้:

*   ช่องชื่อ (ข้อความ) → **string**
*   ช่องอายุ (ตัวเลข) → **integer**
*    อุณหภูมิร่างกาย (ทศนิยม) → **float**
*   สูบบุหรี่หรือไม่? (ใช่/ไม่ใช่) → **boolean**

````python
# Variables store information (like a box with a label)
patient_name = "John Doe"          # Text (string)
age = 45                           # Number (integer)
temperature = 36.5                 # Decimal number (float)
is_smoker = True                   # True/False (boolean)

print(f"Patient: {patient_name}")
print(f"Age: {age}")
print(f"Temperature: {temperature}°C")
print(f"Smoker: {is_smoker}")
````

🔄 **การแปลงชนิดข้อมูล (CONVERTING BETWEEN DATA TYPES)**

ในการเตรียมข้อมูล คุณมักจะได้รับข้อมูลที่อยู่ในรูปแบบที่ไม่ถูกต้องบ่อยๆ:

• อายุถูกจัดเก็บเป็นข้อความ: `"45"` แทนที่จะเป็นตัวเลข `45`

• ตัวเลขมีทศนิยมทั้งๆ ที่คุณต้องการแค่จำนวนเต็ม

• จำเป็นต้องแปลงตัวเลขให้เป็นข้อความเพื่อนำไปสร้างรายงาน

**การแปลงชนิดข้อมูลที่ใช้บ่อย (COMMON CONVERSIONS):**

| ฟังก์ชัน (Function) | แปลงเป็น (Converts To) | ตัวอย่าง (Example) | ผลลัพธ์ (Result) |
| --- | --- | --- | --- |
| **int()** | จำนวนเต็ม (Integer) | `int("45")` | `45` |
| **float()** | ทศนิยม (Decimal) | `float("45.5")` | `45.5` |
| **str()** | ข้อความ (Text/String) | `str(45)` | `"45"` |
| **bool()** | ค่าจริง/เท็จ (True/False) | `bool(1)` | `True` |

**ปัญหาที่พบบ่อยในข้อมูลที่ได้มา:**

1. **อายุถูกอิมพอร์ตเข้ามาเป็นข้อความ:** `"45"` → ทำให้นำไปคำนวณทางสถิติหรือหาค่าเฉลี่ยต่อไม่ได้
2. **ระยะโรคมีจุดทศนิยม:** `2.0` → จริงๆ ควรเป็นจำนวนเต็ม `2`
3. **เลขบัตรประชาชน (CID) เป็นตัวเลข:** `1234567890123` → ต้องแปลงเป็นข้อความ `"1234567890123"` เพื่อป้องกันปัญหาสูญเสียศูนย์ตัวหน้า หรือฟอร์แมตตัวเลขผิดเพี้ยน

**การตรวจสอบชนิดข้อมูล (CHECKING DATA TYPE):**

### 1.2 🧮 BASIC ARITHMETIC OPERATORS


BASIC MATH OPERATORS:

| Operator | Meaning        | Example      | Result |
|----------|----------------|--------------|--------|
| +        | Addition       | 5 + 3        | 8      |
| -        | Subtraction    | 5 - 3        | 2      |
| *        | Multiplication | 5 * 3        | 15     |
| /        | Division       | 10 / 2       | 5.0    |
| //       | Floor Division | 10 // 3      | 3      |
| %        | Remainder      | 10 % 3       | 1      |
| **       | Power          | 2 ** 3       | 8      |
หมายเหตุ
- ตัวดำเนินการ / จะให้ผลลัพธ์เป็น ตัวเลขทศนิยม (Decimal) เสมอ
- ตัวดำเนินการ // จะให้ผลลัพธ์เป็น จำนวนเต็ม (Whole number)


````python
# Addition
total_patients = 50 + 30
print(f"Total patients: 50 + 30 = {total_patients}")

# Subtraction
remaining = 100 - 25
print(f"Remaining: 100 - 25 = {remaining}")

# Multiplication
total_visits = 15 * 4
print(f"Total visits: 15 * 4 = {total_visits}")
````

````python
# Division (always gives decimal)
average = 100 / 4
print(f"Average: 100 / 4 = {average}")

# Floor division (whole number only)
groups = 100 // 3
print(f"Complete groups: 100 // 3 = {groups}")

# Remainder (modulo)
leftover = 100 % 3
print(f"Leftover: 100 % 3 = {leftover}")

# Power
squared = 5 ** 2
print(f"5 squared: 5 ** 2 = {squared}")
````

###1.4 🔍 COMPARISONS & IF STATEMENTS

**1) COMPARISONS**

การเปรียบเทียบ (Comparisons) คือการตรวจสอบว่าเงื่อนไขบางอย่างเป็น **จริง (True)** หรือ **เท็จ (False)** เปรียบเสมือนการตั้งคำถาม เช่น:

• ผู้ป่วยรายนี้อายุมากกว่า 60 ปีหรือไม่?

• ระยะของโรคเท่ากับ 'IV' หรือไม่?

• วันที่วินิจฉัยโรคเกิดขึ้นหลังปี 2020 หรือไม่?

**ตัวดำเนินการเปรียบเทียบ (COMPARISON OPERATORS):**

| ตัวดำเนินการ (Operator) | ความหมาย (Meaning) | ตัวอย่าง (Example) |
| --- | --- | --- |
| `==` | เท่ากับ | `age == 65` |
| `!=` | ไม่เท่ากับ | `sex != 'M'` |
| `>` | มากกว่า | `age > 60` |
| `<` | น้อยกว่า | `age < 50` |
| `>=` | มากกว่าหรือเท่ากับ | `age >= 60` |
| `<=` | น้อยกว่าหรือเท่ากับ | `age <= 50` |

⚠️ **ข้อผิดพลาดที่พบบ่อย (COMMON MISTAKE):**

• `=` ใช้สำหรับการ **กำหนดค่า** (setting a value)
`age = 45` → "กำหนดให้ age มีค่าเท่ากับ 45"

• `==` ใช้สำหรับการ **เปรียบเทียบ** (checking equality)
`age == 45` → "age มีค่าเท่ากับ 45 ใช่หรือไม่?"

**ตัวดำเนินการทางตรรกศาสตร์ (LOGICAL OPERATORS - การรวมหลายเงื่อนไข):**

| ตัวดำเนินการ (Operator) | ความหมาย (Meaning) |
| --- | --- |
| **and** | เงื่อนไข **ต้องเป็นจริงทั้งหมด** |
| **or** | เงื่อนไข **เป็นจริงอย่างน้อยหนึ่งอย่าง** |
| **not** | **กลับค่าผลลัพธ์** (จากจริงเป็นเท็จ / จากเท็จเป็นจริง) |

**ตัวอย่าง:**

• `age > 60 and stage == 'IV'` → ทั้งคู่อายุมากกว่า 60 และระยะโรคต้องเป็น IV จึงจะเป็นจริง

• `age > 60 or stage == 'IV'` → อย่างใดอย่างหนึ่งเป็นจริง (หรือเป็นจริงทั้งคู่) ก็ถือว่าเป็นจริง

• `not is_smoker` → เป็นจริงถ้า `is_smoker` มีค่าเป็น False (ไม่ได้สูบบุหรี่)

**2) คำสั่ง IF (MAKING DECISIONS - การตัดสินใจตามเงื่อนไข):**

**โครงสร้าง:**

```python

    ทำสิ่งนี้

```

**เปรียบเทียบเหมือนผังงาน (Flowchart):**

```text
         อายุ >= 60 หรือไม่?
            /       \
          ใช่       ไม่ใช่
           |          |
      [ผู้สูงอายุ] [ไม่ใช่ผู้สูงอายุ]

```

**คุณจะได้ใช้การเปรียบเทียบเพื่อ:**

• **กรองข้อมูล (Filter data):** "แสดงเฉพาะผู้ป่วยสูงอายุ"

• **ค้นหาข้อผิดพลาด (Find errors):** "หาข้อมูลที่ระบุอายุมากกว่า 120 ปี"

• **จัดกลุ่มข้อมูล (Categorize):** "ถ้าเป็นระยะ I หรือ II ให้ระบุว่าเป็น 'ระยะเริ่มต้น (Early)'"

````python
# 1. การนิยามสร้างฟังก์ชัน (Define Function)
def assess_patient_risk(age, smoking):
    if age >= 65 and smoking:
        return "High Risk (ความเสี่ยงสูง)"
    elif age >= 65 or smoking :
        return "Moderate Risk (ความเสี่ยงปานกลาง)"
    else:
        return "Low Risk (ความเสี่ยงต่ำ)"

# 2. การเรียกใช้งานฟังก์ชัน (Call Function)
# คนไข้คนที่ 1
patient1_result = assess_patient_risk(68, True)
print(f"คนไข้รายที่ 1: {patient1_result}")

# คนไข้คนที่ 2 (รันบรรทัดเดียวจบ ไม่ต้องเขียน if-else ใหม่!)
patient2_result = assess_patient_risk(45, False)
print(f"คนไข้รายที่ 2: {patient2_result}")
````

###2.2  👓 Access and filter dataframe  

บทเรียนนี้สรุปวิธีการเข้าถึงข้อมูลใน Pandas DataFrame แบบปูพื้นฐานพร้อมตัวอย่างการรันโค้ดจริง

````python
import pandas as pd

# 1. สร้าง ข้อมูลตัวอย่าง
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['Bangkok', 'Chiang Mai', 'Phuket', 'Pattaya'],
    'Score': [85.5, 92.0, 78.5, 88.0]
}

# กำหนด Index เป็นชื่อ id เพื่อให้เห็นภาพการอ้างอิงด้วย Index Label
df = pd.DataFrame(data, index=['id_1', 'id_2', 'id_3', 'id_4'])

print("--- DataFrame ตัวอย่าง ---")
display(df)  # ใช้ display() ใน Colab จะแสดงผลเป็นตารางสวยงาม
````

---
2.2.1 การเข้าถึง Column (คอลัมน์)

การเข้าถึงข้อมูลรายคอลัมน์สามารถทำได้ 2 รูปแบบหลัก:
* **เลือกคอลัมน์เดียว:**
  * ใช้ `df['ชื่อคอลัมน์']` (แนะนำ) ผลลัพธ์ที่ได้จะมีชนิดข้อมูลเป็น **Series**
  * หรือใช้ `df.ชื่อคอลัมน์` (Dot Notation) จะใช้ได้เฉพาะชื่อคอลัมน์ที่ไม่ไม่มีเว้นวรรคหรือสัญลักษณ์พิเศษ
* **เลือกหลายคอลัมน์:**
  * ใส่ List ของชื่อคอลัมน์ซ้อนเข้าไป เช่น `df[['Col1', 'Col2']]` ผลลัพธ์ที่ได้จะเป็น **DataFrame**

````python
# 1.1 เลือกคอลัมน์เดียว 'Age' -> ได้ผลลัพธ์เป็น Series
ages = df['Age']
print("--- เลือกคอลัมน์เดียว (df['Age']) ---")
print(ages)
print("\nType ของข้อมูล:", type(ages))

# 1.2 เลือกหลายคอลัมน์ 'Name' และ 'Score' -> ได้ผลลัพธ์เป็น DataFrame
subset = df[['Name', 'Score']]
print("\n--- เลือกหลายคอลัมน์ (df[['Name', 'Score']]) ---")
display(subset)
````

---
2.2.2 การเข้าถึง Row (แถว)

การระบุแถวใน Pandas มีเครื่องมือหลัก 2 ตัวที่ต้องจำให้แม่น:

| เครื่องมือ | หลักการทำงาน | ตัวอย่าง |
| :--- | :--- | :--- |
| **`.loc[]`** | อ้างอิงด้วย **Label (ชื่อ Index)** | `'id_1'`, `'id_2'` |
| **`.iloc[]`** | อ้างอิงด้วย **Integer Position (ลำดับตัวเลข)** | `0`, `1`, `2` (เริ่มนับจาก 0) |


````python
# 2.1 เลือกแถวเดียว
print("--- ดึงแถว 'id_2' ด้วย .loc['id_2'] ---")
display(df.loc['id_2'])

print("\n--- ดึงแถวแรก (ลำดับที่ 0) ด้วย .iloc[0] ---")
display(df.iloc[0])

# 2.2 เลือกหลายแถว
print("\n--- เลือกเฉพาะแถว 'id_1' และ 'id_3' ด้วย .loc ---")
display(df.loc[['id_1', 'id_3']])
````

---
2.2.3 การเข้าถึง Data เฉพาะจุด (Cell-Specific Access)

เมื่อต้องการเข้าถึงค่าในเซลล์เดียว (จุดที่ Row ตัดกับ Column) สามารถทำได้ดังนี้:

* **อ้างอิงด้วยชื่อ:** `df.loc[Row_Label, Column_Name]`
* **อ้างอิงด้วยลำดับตัวเลข:** `df.iloc[Row_Position, Column_Position]`

2.2.4 ✏️ การแก้ไขข้อมูลด้วย `df.at` และ `df.iat`

`df.at` เป็นวิธีที่ถูกออกแบบมาเพื่อ **เข้าถึงและแก้ไขข้อมูลในเซลล์เดียว (Scalar Value)** ที่มีประสิทธิภาพและความเร็วสูงที่สุดใน Pandas


```python
# รูปแบบการใช้
df.at[Row_Label, Column_Name] = ค่าใหม่

````python
# 1 ดึงค่าอายุของ id_2 ด้วยชื่อ
age_bob = df.loc['id_2', 'Age']
print(f"อายุของ id_2 (df.loc['id_2', 'Age']): {age_bob}")

# 2 ดึงค่าอายุของ id_2 ด้วยลำดับตำแหน่ง (แถวที่ 1, คอลัมน์ที่ 1)
age_bob_iloc = df.iloc[1, 1]
print(f"อายุของ id_2 (df.iloc[1, 1]): {age_bob_iloc}")

# 3 ดึงข้อมูลเป็นช่วงตารางย่อย (Sub-table)
print("\n--- ดึงแถว id_1 ถึง id_3 เฉพาะคอลัมน์ Name และ City ---")
display(df.loc['id_1':'id_3', ['Name', 'City']])

# 4 อ่านและแก้ไขข้อมูลด้วย .at[]
print("\n--- แก้ไขค่า Score ของ id_3 ด้วย .at[] ---")
print("คะแนนเดิม:", df.at['id_3', 'Score'])
df.at['id_3', 'Score'] = 80.0
print("คะแนนใหม่หลังแก้ไข:", df.at['id_3', 'Score'])
````

2.2.4 🔍 การเลือกข้อมูลแบบมีเงื่อนไข (Conditional Selection)

หรือเรียกว่า Boolean Indexing คือการใช้เงื่อนไขทางคณิตศาสตร์หรือตรรกศาสตร์ในการกรองเอาเฉพาะแถวที่ต้องการ:

โดยมากใช้ AND (`&`), OR (`|`) และ `.isin()` ใน Pandas

⚠️ สำคัญ

1. **ห้ามใช้คำว่า `and` / `or`** ให้ใช้เครื่องหมายสัญลักษณ์แทน:
   * **`&`** แทน **AND (และ)** → ต้องผ่าน **ทุกเงื่อนไข**
   * **`|`** แทน **OR (หรือ)** → ผ่าน **เงื่อนไขใดเงื่อนไขหนึ่ง** ก็พอ
2. **ต้องใส่เครื่องหมายวงเล็บ `( )` ครอบแต่ละเงื่อนไขเสมอ!**
   * ❌ `df[df['Age'] > 25 & df['Score'] > 80]` *(เกิด Error)*
   * ✅ `df[(df['Age'] > 25) & (df['Score'] > 80)]` *(ถูกต้อง)*

💡 การใช้ `.isin()` ใน Pandas (เช็คค่าในกลุ่มข้อมูล)

เมธอด **`.isin()`** ใช้สำหรับกรองข้อมูลเมื่อต้องการเช็คว่า **"ค่าในคอลัมน์ ตรงกับค่าใดค่าหนึ่งใน List ที่เรากำหนดไว้หรือไม่"** ซึ่งช่วยให้โค้ดสั้นและอ่านง่ายกว่าการใช้เงื่อนไข OR (`|`) หลายๆ ตัวต่อกัน

⚖️ เปรียบเทียบ: ใช้ `|` vs ใช้ `.isin()`

สมมติว่าต้องการหาคนทีอยู่ในเมือง **'Bangkok'**, **'Phuket'** หรือ **'Pattaya'**

❌ แบบเดิม (ใช้ `|` ต่อกันหลายตัว):
โค้ดยาว อ่านยาก และต้องใส่วงเล็บหลายชั้น
```python
df[(df['City'] == 'Bangkok') | (df['City'] == 'Phuket') | (df['City'] == 'Pattaya')]
```
✅ ใช้ isin ทำให้ code กระชับ และอ่านง่ายขึ้น
```python
df[df['City'].isin(['Bangkok', 'Phuket', 'Pattaya'])]
```
✅ ใช้ `&` ร่วมกับ `.isin()` (เช็คกลุ่มข้อมูล "และ" เงื่อนไขอื่น)
```python
# หาคนที่อยู่ในเมือง 'Bangkok' หรือ 'Phuket' "และ" มี Score มากกว่า 80
df[(df['City'].isin(['Bangkok', 'Phuket'])) & (df['Score'] > 80)]

````python
# 4.1 กรองเอาเฉพาะแถวที่ Age มากกว่า 30
print("--- แสดงข้อมูลของคนที่ Age > 30 ---")
display(df[df['Age'] > 30])

# 4.2 กรองเอาคนที่มี City เป็น 'Bangkok' และเลือกแสดงเฉพาะคอลัมน์ Name และ Score
print("\n--- แสดงเฉพาะ Name และ Score ของคนที่อยู่ Bangkok ---")
display(df.loc[df['City'] == 'Bangkok', ['Name', 'Score']])

# หาคนที่อยู่ในเมือง 'Bangkok' หรือ 'Phuket' "และ" มี Score มากกว่า 80
print("\n--- แสดงคนที่อยู่ในเมือง 'Bangkok' หรือ 'Phuket' และ มี Score มากกว่า 80")
display(df[(df['City'].isin(['Bangkok', 'Phuket'])) & (df['Score'] > 80)])
````
